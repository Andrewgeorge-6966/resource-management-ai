```python
import pandas as pd


# 2️⃣ Read the task summary dataset
task_summary_df = pd.read_csv("CESAW_task_fact.csv")  # replace with your file path
print("\nTask Summary Dataset:")
print(task_summary_df.head())
print(task_summary_df.info())

# 3️⃣ Read the WBS hierarchy dataset
wbs_df = pd.read_csv("wbs_parent.csv")  # replace with your file path
print("\nWBS Dataset:")
print(wbs_df.head())
print(wbs_df.info())
```

```python
df = pd.merge(
    task_summary_df,
    wbs_df,
    on=['project_key', 'wbs_element_key'],
    how='left'
)

print("Task + WBS Merge:")
print(df.head())
```
